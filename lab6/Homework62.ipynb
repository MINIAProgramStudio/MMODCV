```python
import cv2
```

### Step 1
Decide what video you are going to use for this homework, select an object and generate the template. You can use any video you want (your own, from Youtube, etc.)
and track any object you want (e.g. a car, a pedestrian, etc.).


### Step 2
Initialize a tracker (e.g. KCF).

```python
video_path = "UP4014Short.mp4"
frames_to_track = 30
save_each_N_frames = 3

cap = cv2.VideoCapture(video_path)

ret, frame = cap.read()
if not ret:
    raise RuntimeError("Could not read video")

bbox_start = cv2.selectROI("Frame - Select ROI", frame, False)
cv2.destroyWindow("Frame - Select ROI")

tracker1 = cv2.TrackerKCF_create()
tracker1.init(frame, bbox_start)

```

### Step 3
Run the tracker on the video and the selected object. Run the tracker for around 10-15 frames.

### Step 4
For each frame, print the bounding box on the image and save it.

```python
frame_id = 0

while frame_id < frames_to_track:
    ret, frame = cap.read()
    if not ret:
        break

    ok, bbox = tracker1.update(frame)

    if ok:
        x, y, w, h = map(int, bbox)
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
    if frame_id % save_each_N_frames == 0:
        cv2.imwrite(f"frames/kcf_{frame_id:04d}.jpg", frame)
    frame_id += 1

cap.release()
```

### Step 5
Select a different tracker (e.g. CSRT) and repeat steps 2, 3 and 4.

```python
cap = cv2.VideoCapture(video_path)

ret, frame = cap.read()
if not ret:
    raise RuntimeError("Could not read video")

tracker2 = cv2.TrackerCSRT_create()
tracker2.init(frame, bbox_start)

frame_id = 0

while frame_id < frames_to_track:
    ret, frame = cap.read()
    if not ret:
        break

    ok, bbox = tracker2.update(frame)

    if ok:
        x, y, w, h = map(int, bbox)
        cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)
    if frame_id % save_each_N_frames == 0:
        cv2.imwrite(f"frames/csrt_{frame_id:04d}.jpg", frame)
    frame_id += 1

cap.release()
```
